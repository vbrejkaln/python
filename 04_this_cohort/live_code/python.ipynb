SET OPERATIONS

```python
rainbow = {'red', 'orange', 'yellow', 'green', 'blue', 'indigo', 'violet'}
olympic_flag = {'red', 'yellow', 'green', 'blue', 'black'}

rainbow.difference(olympic_flag) #returns a set of items in rainbow that are not in olympic_flag

```

```python
rainbow.symmetric_difference(olympic_flag) #returns a set of items that are in either rainbow or olympic_flag but not both
```

```python
olympic_flag.difference(rainbow) #returns a set of items in olympic_flag that are not in rainbow
```

```python
olympic_flag.symmetric_difference(rainbow) #returns a set of items that are in either olympic_flag or rainbow but not both
```

```python
rainbow.intersection(olympic_flag) #returns a set of items in both rainbow and olympic_flag
```

```python
rainbow.union(olympic_flag) #returns a set of items in either rainbow or olympic_flag
```
